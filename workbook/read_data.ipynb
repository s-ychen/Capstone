```python
from pathlib import Path
import pandas as pd
from IPython.display import display

import matplotlib.pyplot as plt

# Find Parquet files under the notebook's working directory.
parquet_files = sorted(Path.cwd().rglob("*.parquet"))

if not parquet_files:
    print(f"No Parquet files found under {Path.cwd()}")
else:
    parquet_tables = {}

    for path in parquet_files:
        try:
            df = pd.read_parquet(path)
            parquet_tables[str(path)] = df
            print(f"\n{path} — {df.shape[0]:,} rows × {df.shape[1]} columns")
            display(df.head())

            numeric = df.select_dtypes(include="number")
            if not numeric.empty:
                numeric.hist(figsize=(10, 6), bins=30)
                plt.suptitle(path.name)
                plt.tight_layout()
                plt.show()
        except Exception as exc:
            print(f"Could not read {path}: {exc}")
```
